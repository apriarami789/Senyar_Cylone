```python
# @title Import Libraray
from pathlib import Path
import os
import glob

import numpy as np
import pandas as pd
import xarray as xr
import polars as pl
import matplotlib

import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from matplotlib.animation import FuncAnimation
from IPython.display import display, HTML, Image, Markdown, Video
```

```python
# @title List of Senyar_Cylone folder file
# pemilihan workspace
path_top = Path('/content/Senyar_Cylone/')
os.listdir(path_top)
```

```python
# open Single Level data
ds_single = xr.open_dataset(path_top / 'ERA5_SingleLevel_DailyMeanandSum_20250101-20260331.nc', engine='netcdf4')
display(ds_single)
```

```python
# @title
# Mapping manual atau menggunakan metadata dari ds_single
# Kita coba ambil 'long_name' dan 'units' langsung dari atribut dataset jika ada,
# serta menyediakan fallback deskripsi bahasa Indonesia yang rapi dalam format LaTeX.

variable_meanings = {
    'pv': 'Potential Vorticity (Vortisitas Potensial)',
    'crwc': 'Specific rain water content (Kandungan air hujan spesifik)',
    't': 'Temperature (Suhu)',
    'q': 'Specific humidity (Kelembapan spesifik)',
    'w': 'Vertical velocity (Kecepatan vertikal)',
    'r': 'Relative humidity (Kelembapan relatif)',
    'd': 'Divergence (Divergensi)',
    'clwc': 'Specific cloud liquid water content (Kandungan air awan cair spesifik)',
    'vo': 'Relative vorticity (Vortisitas relatif)',
    'cc': 'Fraction of cloud cover (Fraksi tutupan awan)',
    'v': 'V-component of wind (Komponen angin Utara-Selatan)',
    'u': 'U-component of wind (Komponen angin Barat-Timur)'
}

markdown_text = "### Daftar Variabel Dataset dan Satuannya\n\n"
markdown_text += "Berikut adalah daftar variabel yang terdapat pada dataset beserta penjelasan dan satuannya:\n\n"
markdown_text += "| Nama Variabel | Deskripsi / Kepanjangan | Satuan (LaTeX) |\n"
markdown_text += "| :---: | :--- | :---: |\n"

for var in ds_single.data_vars:
    # Ambil long_name dari metadata netCDF jika ada, jika tidak pakai dictionary
    long_name = ds_single[var].attrs.get('long_name', variable_meanings.get(var, var))

    # Ambil unit asli
    unit = ds_single[var].attrs.get('units', '-')

    # Konversi beberapa unit populer ke format LaTeX
    latex_unit = unit
    if unit == 'K':
        latex_unit = r"$\text{K}$"
    elif 'm s**-1' in unit or 'm/s' in unit:
        latex_unit = r"$\text{m} \cdot \text{s}^{-1}$"
    elif 'kg kg**-1' in unit or 'kg/kg' in unit:
        latex_unit = r"$\text{kg} \cdot \text{kg}^{-1}$"
    elif 'Pa s**-1' in unit:
        latex_unit = r"$\text{Pa} \cdot \text{s}^{-1}$"
    elif 's**-1' in unit:
        latex_unit = r"$\text{s}^{-1}$"
    elif 'K m**2 kg**-1 s**-1' in unit:
        latex_unit = r"$\text{K} \cdot \text{m}^2 \cdot \text{kg}^{-1} \cdot \text{s}^{-1}$"
    elif unit != '-':
        latex_unit = f"$\\text{{{unit}}}$"

    markdown_text += f"| **{var}** | {long_name} | {latex_unit} |\n"

display(Markdown(markdown_text))
```

```python
# @title Membuat fungsi Plot video dalam bentuk .mp4
def vid_analysis(variable_data=None, saving_path=None):
  if variable_data == None and saving_path == None:
    raise ValueError ("Nothing data Variable in and saving")

  # Memastikan limit ukuran animasi diperbesar jika diperlukan
  matplotlib.rcParams['animation.embed_limit'] = 50.0

  # Mengambil variabel skin temperature (skt)
  #data_skt = ds_single['skt']
  data = ds_single[variable_data]

  # Membuat figure dan axes

  # Menyiapkan figure dan axes dengan ruang yang pas (rapat dan rapi)
  fig = plt.figure(figsize=(13, 8))
  ax = plt.axes([0.12, 0.18, 0.74, 0.75], projection=ccrs.PlateCarree())

  # Tentukan batas minimum dan maksimum absolut dari data agar rentang warna konsisten
  vmin = float(data.min().values)
  vmax = float(data.max().values)
  # Membuat tingkat interpolasi yang rapat (halus)
  levels = np.linspace(vmin, vmax, 100)

  # Menentukan 8 ticks angka bulat tanpa koma
  ticks_values = np.linspace(int(np.ceil(vmin)), int(np.floor(vmax)), 8)
  ticks_values = np.round(ticks_values).astype(int)

  # Plot awal (dummy) untuk menginisialisasi colorbar statis
  sample_data = data.isel(valid_time=0)
  im_dummy = ax.contourf(
      sample_data.longitude,
      sample_data.latitude,
      sample_data.values,
      levels=levels,
      transform=ccrs.PlateCarree(),
      cmap='RdBu_r',
      extend='both'
  )

  # Menambahkan colorbar statis secara vertikal di kanan secara presisi dan rapi
  cax = fig.add_axes([0.88, 0.18, 0.03, 0.75])
  cbar = fig.colorbar(im_dummy, cax=cax, orientation='vertical', ticks=ticks_values)
  cbar.set_label('Skin Temperature (K)', fontsize=11, fontweight='bold', labelpad=15)
  cbar.ax.set_yticklabels([str(int(t)) for t in ticks_values], fontsize=10)

  # Fungsi update untuk setiap frame animasi harian
  def update(frame):
      ax.clear()

      # Mengambil data pada timestep tertentu
      step_data = data.isel(valid_time=frame)

      # Plot Contourf untuk interpolasi warna yang halus
      im = ax.contourf(
          step_data.longitude,
          step_data.latitude,
          step_data.values,
          levels=levels,
          transform=ccrs.PlateCarree(),
          cmap='RdBu_r',
          extend='both',
          zorder=1
      )

      # Menambahkan ulang fitur peta setelah ax.clear()
      ax.add_feature(cfeature.COASTLINE, linewidth=1.5, edgecolor='black', zorder=2)
      ax.add_feature(cfeature.BORDERS, linestyle=':', linewidth=1, zorder=2)
      ax.add_feature(cfeature.LAND, facecolor='none', edgecolor='black', alpha=0.1, zorder=1)
      ax.add_feature(cfeature.LAKES, alpha=0.6, zorder=2)

      # Menambahkan gridlines
      gl = ax.gridlines(draw_labels=True, alpha=0.3, linestyle='--', zorder=3)
      gl.top_labels = False
      gl.right_labels = False

      # Menambahkan label sumbu secara eksplisit dan presisi di bagian kiri dan bawah peta
      fig.text(0.49, 0.11, 'Longitude', ha='center', va='center', fontsize=12, fontweight='bold')
      fig.text(0.09, 0.55, 'Latitude', ha='center', va='center', rotation='vertical', fontsize=12, fontweight='bold')

      # Update judul dengan format tanggal
      time_str = str(step_data.valid_time.values)[:10]
      ax.set_title(f"Daily Smoothed Skin Temperature - {time_str}", fontsize=14, pad=15, fontweight='bold')
      return [im]

  # Membuat objek animasi
  anim = FuncAnimation(fig, update, frames=len(data.valid_time), interval=300, blit=False)

  # Menyimpan animasi ke format .mp4 menggunakan ffmpeg
  output_filename = saving_path
  #output_filename = path_top / 'skin_temperature_daily.mp4'
  try:
      print("Sedang memproses dan menyimpan video .mp4 dengan warna halus...")
      anim.save(output_filename, writer='ffmpeg', fps=3, dpi=150)
      print(f"Berhasil! Video halus disimpan dengan nama: {output_filename}")

      # Menampilkan ukuran file .mp4 yang dibuat
      file_size_bytes = os.path.getsize(output_filename)
      file_size_mb = file_size_bytes / (1024 * 1024)
      print(f"Ukuran file video: {file_size_mb:.2f} MB")
  except Exception as e:
      print(f"Terjadi kesalahan saat menyimpan file .mp4: {e}")

  plt.close()
```

```python
# membuat video dari variabel skt
vid_analysis(variable_data='skt', saving_path=path_top / 'skin_temperature_daily.mp4')
```
